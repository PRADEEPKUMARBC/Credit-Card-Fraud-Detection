```python
# Cell 8: Cleaning and feature/target separation

# Reload a fresh copy so this cell is idempotent
df_raw = load_dataset()

print("Raw shape:", df_raw.shape)

# ---------------------------------------------------------
# 1. Clean the dataset
# ---------------------------------------------------------

df_clean = df_raw.copy()

# Check duplicates
duplicates_before = df_clean.duplicated().sum()

print("\nDuplicate rows found:", duplicates_before)

# Remove duplicate rows
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("Cleaned shape:", df_clean.shape)
print("Duplicates removed:", duplicates_before)

# ---------------------------------------------------------
# 2. Class balance after cleaning
# ---------------------------------------------------------

print("\nClass balance after cleaning:")

class_counts = df_clean["Class"].value_counts().sort_index()
class_percentages = (
    df_clean["Class"].value_counts(normalize=True).sort_index() * 100
)

class_balance = pd.DataFrame({
    "Count": class_counts,
    "Percentage": class_percentages.round(4)
})

class_balance.index = ["Genuine (0)", "Fraudulent (1)"]

print(class_balance)

# ---------------------------------------------------------
# 3. Separate features and target
# ---------------------------------------------------------

X = df_clean.drop(columns=["Class"])
y = df_clean["Class"]

print("\nFeature shape:", X.shape)
print("Target shape :", y.shape)

print("\nTarget classes:")
print(sorted(y.unique()))

# ---------------------------------------------------------
# 4. Define columns to scale
# ---------------------------------------------------------

cols_to_scale = ["Time", "Amount"]

# Make sure the columns exist
cols_to_scale = [
    col for col in cols_to_scale
    if col in X.columns
]

print("\nColumns to scale:", cols_to_scale)

# ---------------------------------------------------------
# 5. Sample ranges before scaling
# ---------------------------------------------------------

print("\nSample ranges (before scaling):")

print(
    X[cols_to_scale]
    .describe()
    .T[["min", "max", "mean", "std"]]
)
```

```python
# Cell 9: Split, scale, and apply SMOTE

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

RANDOM_STATE = 42
TEST_SIZE = 0.2

# ---------------------------------------------------------
# 1. Train/Test Split
# ---------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Train/Test Split")
print("-----------------")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print("\nClass distribution before SMOTE:")
print(y_train.value_counts().sort_index())


# ---------------------------------------------------------
# 2. Scale Time and Amount
# ---------------------------------------------------------

scaled_cols = ["Time", "Amount"]

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

# Fit scaler ONLY on training data
X_train_scaled[scaled_cols] = scaler.fit_transform(
    X_train[scaled_cols]
)

# Apply the same scaler to test data
X_test_scaled[scaled_cols] = scaler.transform(
    X_test[scaled_cols]
)

print("\nScaling completed.")
print("Scaled columns:", scaled_cols)


# ---------------------------------------------------------
# 3. Apply SMOTE ONLY to training data
# ---------------------------------------------------------

smote = SMOTE(
    random_state=RANDOM_STATE,
    k_neighbors=5
)

X_train_bal, y_train_bal = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("\nSMOTE Results")
print("-------------")

print("Before SMOTE:")
print(y_train.value_counts().sort_index())

print("\nAfter SMOTE:")
print(y_train_bal.value_counts().sort_index())


# ---------------------------------------------------------
# 4. Final shapes
# ---------------------------------------------------------

print("\nFinal shapes:")
print("X_train_bal  :", X_train_bal.shape)
print("y_train_bal  :", y_train_bal.shape)
print("X_test_scaled:", X_test_scaled.shape)
print("y_test       :", y_test.shape)
```

```python
# Cell 10: Visualize class distribution before vs after SMOTE

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# ---------------------------------------------------------
# Before SMOTE
# ---------------------------------------------------------

before = y_train.value_counts().sort_index()

axes[0].bar(
    ["Genuine", "Fraud"],
    before.values,
    color=["#2E86AB", "#E63946"],
    edgecolor="black"
)

axes[0].set_title("Training Set — Before SMOTE")
axes[0].set_ylabel("Number of Transactions")
axes[0].set_yscale("log")

for i, value in enumerate(before.values):
    axes[0].text(
        i,
        value,
        f"{value:,}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )


# ---------------------------------------------------------
# After SMOTE
# ---------------------------------------------------------

after = y_train_bal.value_counts().sort_index()

axes[1].bar(
    ["Genuine", "Fraud"],
    after.values,
    color=["#2E86AB", "#E63946"],
    edgecolor="black"
)

axes[1].set_title("Training Set — After SMOTE")
axes[1].set_ylabel("Number of Transactions")

for i, value in enumerate(after.values):
    axes[1].text(
        i,
        value,
        f"{value:,}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )


plt.suptitle(
    "Effect of SMOTE on Training Class Balance",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

# Save figure
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

output_path = REPORTS_DIR / "smote_comparison.png"

plt.savefig(
    output_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(f"Saved: {output_path}")
```
